```python
import pdfplumber
```

```python
def extract_resume_text(pdf_path):

    text = ""

    with pdfplumber.open(pdf_path) as pdf:

        for page in pdf.pages:
            text += page.extract_text()

    return text.lower()
```

```python
skill_keywords = {
    'Skills - Python': ['python'],
    'Skills - SQL': ['sql', 'mysql', 'postgresql'],
    'Skills - ML': ['machine learning', 'scikit-learn'],
    'Skills-DeepLearning': ['tensorflow', 'pytorch', 'deep learning'],
    'skills_Cloud': ['aws', 'azure', 'gcp', 'cloud']
}
```

```python
def extract_skills(resume_text):

    extracted_skills = {}

    for skill, keywords in skill_keywords.items():

        found = 0

        for keyword in keywords:

            if keyword in resume_text:
                found = 1
                break

        extracted_skills[skill] = found

    return extracted_skills
```

```python
resume_text = extract_resume_text("sample_resume.pdf")

skills = extract_skills(resume_text)

print(skills)
```
