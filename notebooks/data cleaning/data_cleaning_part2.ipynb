Salary (USD) :
1. Completeness issues
2. validation issues like : many salaries is negative
3. some salary existed with special chars and alphabets
4. Outliers exist
5. accuracy issue - in some records salary is mentioned as 0.0

```python
df['Salary (USD)'].isnull().sum()
```

```python
df['Salary (USD)'].unique()
```

```python
df['Salary (USD)'] = df['Salary (USD)'].str.replace('$',' ').str.strip().str.replace(',','')
```

```python
def convert_salary_to_float(salary_str):
    if pd.isna(salary_str):
        return np.nan
    s = str(salary_str).lower().strip()
    if 'k' in s:
        s = s.replace('k', '')
        try:
            return float(s) * 1000
        except ValueError:
            return np.nan
    else:
        try:
            return float(s)
        except ValueError:
            return np.nan

df['Salary (USD)'] = df['Salary (USD)'].apply(convert_salary_to_float)
```

```python
def negative_to_positive(price):
  if price < 0:
    return abs(price)
  elif price == np.nan :
    return np.nan
  else:
    return price
```

```python
df['Salary (USD)'] = df['Salary (USD)'].apply(negative_to_positive)
```

```python
Q1 = df['Salary (USD)'].quantile(0.25)
Q3 = df['Salary (USD)'].quantile(0.75)
IQR = Q3 -Q1

lower_value = Q1 - (1.5 * IQR)
print('lower_value', lower_value)
upper_value = Q3 + (1.5 * IQR)
print('upper_value', upper_value)

min = df['Salary (USD)'].quantile(0.05)
lower_value = max(lower_value, min)
print('new', lower_value)
```

```python
df['Salary (USD)'] = df['Salary (USD)'].clip(lower=lower_value, upper=upper_value)
```

```python
df['Salary (USD)'].fillna(df['Salary (USD)'].median(), inplace=True)
```

```python
df['Job Openings'] = df['Job Openings'].replace(['UNKNOWN', 'na'], np.nan)
df['Job Openings'] = pd.to_numeric(df['Job Openings'], errors='coerce')
df['Job Openings'] = df['Job Openings'].fillna(0)
```

```python
df['Job Openings'].isnull().sum()
```

```python
import pandas as pd
import numpy as np
def negetive_to_positive(num):
  # Check if num is a numeric type (int or float) and is negative
  if isinstance(num, (int, float)) and num < 0:
    return abs(num)
  # Use pd.isna() for robust NaN checking
  elif pd.isna(num):
    return np.nan
  else:
    return num
```

```python
df['Job Openings'] = df['Job Openings'].apply(negetive_to_positive)
```

```python
df['Job Openings'] = df['Job Openings'].astype(int)
```
