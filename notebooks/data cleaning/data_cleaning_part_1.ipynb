Years_Experience -


1.  completeness issue
2.   validity - in some rows experience mentioned in negetive (-8, -9)
3. accuracy - in some records mentioned experience is 999 years
4. mixed - numerical + strings in some records



```python
df['Years_Experience'].isnull().sum()
```

```python
df['Years_Experience'] = pd.to_numeric(df['Years_Experience'], errors='coerce')
df['Years_Experience'] = df['Years_Experience'].apply(lambda x: abs(x) if pd.notnull(x) and x < 0 else x)
```

```python
df['Years_Experience'] = df['Years_Experience'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
```

```python
df['Years_Experience'] = df['Years_Experience'].fillna(df['Years_Experience'].median())
```

```python
df['Years_Experience'] = df['Years_Experience'].replace(999.0, df['Years_Experience'].median())
```
