# **Summary of Dataset**

**Column Description**

1. Job ID -

A unique identifier assigned to each job listing.
It helps in distinguishing one record from another and is especially useful when handling duplicates, merging datasets, or tracking specific job entries.


2. JobTitle -

Represents the role being offered (e.g., Data Scientist, ML Engineer, AI Researcher). This column is crucial for:
Understanding demand for specific roles,
Categorizing jobs,
Performing role-based salary or skill analysis.


3. Company Size -

Indicates the scale of the organization (e.g., Small, Medium, Large).
This helps analyze:
Salary differences by company size,
Hiring trends across startups vs enterprises,
Resource availability and hiring capacity.


4. company_Industry -

Specifies the industry sector (e.g., Tech, Finance, Healthcare).
Useful for:
Identifying which industries are investing most in AI/ML,
Industry-wise salary comparisons,
Domain-specific hiring trends.


5. Country -

Represents the geographical location of the job.
Important for:
Regional job demand analysis,
Country-wise salary comparison,
Understanding global hiring trends.


6. Remote Type -

Indicates the work mode:
Remote,
Hybrid,
On-site.
This is valuable for:
Analyzing post-pandemic work trends,
Comparing salaries across work types,
Understanding remote work adoption.

7. Exp. Level -

Categorizes job roles based on experience level:
Entry-level,
Mid-level,
Senior-level.
Useful for:
Career progression insights,
Salary growth analysis,
Skill requirement evolution.

8. Years_Experience -

Represents the exact number of years required for the job.
This allows:
More granular analysis than Exp. Level,
Predictive modeling for salary,
Skill vs experience correlation.

9. Education Level -

Specifies minimum educational qualification (e.g., Bachelor’s, Master’s, PhD).
Understanding academic expectations,
Analyzing salary vs education,
Identifying industry preferences.

Skills Columns

10. Skills - Python

Indicates whether Python is required (binary or categorical).
Important for:
Identifying core programming demand,
Skill frequency analysis.

11. Skills - SQL

Shows requirement of SQL skills.
Used for:
Data-related role analysis,
Backend/data engineering insights.

12. Skills - ML

Represents requirement of Machine Learning knowledge.
Helps track:
ML adoption across roles,
AI-related job demand.

13. Skills - DeepLearning

Indicates whether deep learning is required.
Useful for:
Advanced AI job trend analysis,
Identifying high-skill roles.

14. skills_Cloud

Shows requirement of cloud platforms (AWS, Azure, GCP).
Important for:
Modern infrastructure demand,
DevOps + ML integration insights.

Compensation & Time

15. Salary (USD) -

Represents the salary offered for the role in USD.
This is one of the most critical columns for:
Salary distribution analysis,
Predictive modeling,
Market benchmarking.


16. Posting Month -
Indicates the month when the job was posted.
Useful for:
Seasonal hiring trends,
Monthly job demand patterns.


17. posting_Year -
Represents the year of job posting.
Helps in:
Year-over-year trend analysis,
Market growth tracking.


Hiring Insights

18. Hiring Urgency -
Shows how urgently the company wants to fill the role (e.g., Low, Medium, High).
Useful for:
Demand intensity analysis,
Fast-growing sectors identification.

19. Job Openings -
Represents the number of positions available for a role.
Helps in:
Demand volume estimation,
Company hiring scale analysis.


Metadata Columns

20. data_source -

Indicates where the job listing was collected from (e.g., LinkedIn, Indeed).
Useful for:
Data reliability assessment,
Platform comparison.


21. notes -
Additional comments or extracted insights related to the job.


22. _temp -
a temporary or intermediate column created during data processing.
